# Respiratory Disease Classification — MobileNetV3-Large (scratch + ImageNet-pretrained) on Log-Mel (ICBHI) — ALL 8 optimizers in one notebook

Runs **8 optimizers** — Adam, AdamW, SGD (momentum=0.9), RMSprop, NAdam, RAdam, Adamax, Adagrad — with **two MobileNetV3-Large variants**: `scratch` (random init) and `pretrained` (ImageNet weights, full fine-tuning),
each with 2 learning rates (0.001, 0.0001) × 5 seeds (42, 123, 2024, 7, 2025) = **20 runs per optimizer, 160 runs in total**.
Preprocessing, features, patient-level split (`split_seed=42`), batch size 32, max 150 epochs, patience 15 and the metrics are identical to the ResNet-34 notebook, so results are directly comparable.
All optimizers use PyTorch defaults except the learning rate. No class weighting, oversampling or augmentation.

**Pretrained weights:** the notebook first tries the torchvision download (needs *Internet* enabled in the Kaggle notebook settings). If that fails it looks for a `mobilenet_v3_large*.pth` file in `/kaggle/input`
(add a dataset containing it). It never silently falls back to random weights. Tip: run once with `RESP_SMOKE_TEST=1` (2 epochs, 1 seed, 1 LR) to confirm everything works before the full run.
(Set `mobilenet_size="small"` in `CFG` to use MobileNetV3-Small instead of Large.)

**Outputs (only the main things):**
- `/kaggle/working/mobilenetv3_csv/` — **8 CSV files, one per optimizer** (`mobilenetv3_<optimizer>_results.csv`, one row per run).
- `/kaggle/working/mobilenetv3_download_5parts/` — a zip (the 8 CSVs + all main figures) **split into 5 parts** for easy download, plus rejoin instructions.
- No model checkpoints are saved; the feature cache and per-run files live in a temp folder outside `/kaggle/working`.

Stem conv is changed to 1 input channel (pretrained: RGB filters summed); the classifier keeps MobileNetV3's hidden layer (`Linear → Hardswish`), uses `Dropout(0.5)` and a new final `Linear(hidden, 8)`.
**Time note:** 160 runs is long (MobileNetV3 is much faster per epoch than ResNet-34, but keep the session alive until the end — the temp folder is lost if the session restarts).

```python
# ============================================================
# 4. DIAGNOSIS CSV INSPECTION
# ============================================================
print("--- first 5 raw lines of the file ---")
with open(DIAG_CSV, "r") as f:
    for _ in range(5):
        print(repr(f.readline()))

# The ICBHI file normally has NO header (two columns: patient id, diagnosis). Detect instead of assuming.
raw = pd.read_csv(DIAG_CSV, header=None)
first_cell = str(raw.iloc[0, 0]).strip()
if not first_cell.isdigit():
    print("\nFirst cell is not numeric -> treating first row as a header.")
    raw = pd.read_csv(DIAG_CSV, header=0)
else:
    print("\nNo header detected (first cell is numeric).")
print("Shape:", raw.shape, "| columns:", list(raw.columns))
assert raw.shape[1] >= 2, "Expected at least two columns (patient id, diagnosis)."

diag = raw.iloc[:, :2].copy()
diag.columns = ["patient_id", "diagnosis_raw"]
diag["patient_id"] = pd.to_numeric(diag["patient_id"], errors="coerce")
bad_ids = diag[diag["patient_id"].isna()]
if len(bad_ids):
    print(f"WARNING: {len(bad_ids)} diagnosis rows have a non-numeric patient id and are excluded:\n", bad_ids)
diag = diag.dropna(subset=["patient_id"]).copy()
diag["patient_id"] = diag["patient_id"].astype(int)
diag["diagnosis_raw"] = diag["diagnosis_raw"].astype(str).str.strip()

dups = diag[diag.duplicated("patient_id", keep=False)]
if len(dups):
    conflicting = dups.groupby("patient_id")["diagnosis_raw"].nunique()
    if (conflicting > 1).any():
        raise ValueError(f"Patients with conflicting diagnoses: {list(conflicting[conflicting > 1].index)}")
    print(f"Note: {dups['patient_id'].nunique()} duplicated patient ids with identical diagnoses -> de-duplicated.")
    diag = diag.drop_duplicates("patient_id")

diag["label"] = diag["diagnosis_raw"].str.lower().map(LABEL_MAP)
print("\nPatients in CSV:", len(diag))
print("\nRaw diagnosis values (patients per value):")
print(diag["diagnosis_raw"].value_counts().to_string())

unmapped = diag[diag["label"].isna()]
if len(unmapped):
    print(f"\nWARNING: {len(unmapped)} patients have diagnosis values outside the 8 target classes "
          f"({sorted(unmapped['diagnosis_raw'].unique())}). Their recordings will be excluded and logged.")
missing_classes = [c for c in CLASS_NAMES if c not in set(diag["label"].dropna())]
if missing_classes:
    raise RuntimeError(f"Dataset does not contain target classes: {missing_classes}. Stopping instead of inventing labels.")
print("\nAll 8 target classes are present in the diagnosis CSV.")
```

## 6. Audio Preprocessing

```python
# ============================================================
# 6. AUDIO PREPROCESSING
# ============================================================
TARGET_LEN = int(CFG["sr"] * CFG["duration"])   # 80,000 samples

def load_and_preprocess(path: str) -> np.ndarray:
    """Load -> mono -> resample to 16 kHz -> remove DC offset -> peak-normalise -> pad/truncate to 5 s.
    Raises ValueError for unreadable, empty, non-finite or silent audio (caller logs the reason)."""
    y, _ = librosa.load(path, sr=CFG["sr"], mono=True)           # librosa resamples to CFG['sr']
    if y.size == 0:
        raise ValueError("empty audio")
    if not np.all(np.isfinite(y)):
        raise ValueError("non-finite samples")
    y = y - np.mean(y)                                            # remove DC offset
    peak = np.max(np.abs(y))
    if peak < 1e-8:
        raise ValueError("silent audio")
    y = y / peak                                                  # peak normalisation to [-1, 1]
    if y.size >= TARGET_LEN:                                      # truncate (keep the first 5 s)
        y = y[:TARGET_LEN]
    else:                                                         # zero-pad at the end
        y = np.pad(y, (0, TARGET_LEN - y.size))
    return y.astype(np.float32)

# quick sanity check on one file
_y = load_and_preprocess(meta.loc[0, "path"])
print("Example:", meta.loc[0, "file"], "| shape:", _y.shape, "| min/max:", _y.min(), _y.max(), "| dtype:", _y.dtype)
```

## 7. Log-Mel Spectrogram Generation

```python
# ============================================================
# 7. LOG-MEL SPECTROGRAM GENERATION  (cached to disk)
# ============================================================
def logmel(y: np.ndarray) -> np.ndarray:
    """Power mel spectrogram -> decibel scale. Output shape: (n_mels, frames) = (128, 313)."""
    S = librosa.feature.melspectrogram(y=y, sr=CFG["sr"], n_fft=CFG["n_fft"], hop_length=CFG["hop_length"],
                                       n_mels=CFG["n_mels"], fmin=CFG["fmin"], fmax=CFG["fmax"], power=2.0)
    return librosa.power_to_db(S, ref=np.max, top_db=CFG["top_db"]).astype(np.float32)

# cache key depends on the feature config and the exact file list -> stale caches are never reused
feat_cfg = {k: CFG[k] for k in ["sr", "duration", "n_mels", "n_fft", "hop_length", "fmin", "fmax", "top_db"]}
sig = hashlib.md5((json.dumps(feat_cfg, sort_keys=True) + "|".join(meta["file"])).encode()).hexdigest()[:10]
CACHE_FILE = CACHE_DIR / f"logmel_{sig}.npz"
DROP_FILE = CACHE_DIR / f"dropped_{sig}.json"

if CACHE_FILE.exists():
    z = np.load(CACHE_FILE, allow_pickle=False)
    X_all, good_files = z["X"], list(z["files"])
    dropped_log = json.load(open(DROP_FILE))
    print(f"Loaded cached features: {X_all.shape} from {CACHE_FILE.name}")
else:
    feats, good_files = [], []
    for f, p in tqdm(list(zip(meta["file"], meta["path"])), desc="Log-Mel extraction"):
        try:
            feats.append(logmel(load_and_preprocess(p)))
            good_files.append(f)
        except Exception as e:                                    # corrupted / unreadable / silent file
            dropped_log.append(dict(file=f, reason=f"audio error: {type(e).__name__}: {e}"))
    X_all = np.stack(feats).astype(np.float32)
    np.savez(CACHE_FILE, X=X_all, files=np.array(good_files))
    json.dump(dropped_log, open(DROP_FILE, "w"))
    print(f"Computed and cached features: {X_all.shape} -> {CACHE_FILE.name}")

# keep metadata aligned with the feature array
meta = meta.set_index("file").loc[good_files].reset_index()
assert len(meta) == len(X_all)
dropped_df = pd.DataFrame(dropped_log, columns=["file", "reason"])
dropped_df.to_csv(WORK_RES / "excluded_recordings.csv", index=False)
print(f"\nUsable recordings: {len(meta)} | excluded: {len(dropped_df)}")
if len(dropped_df):
    print(dropped_df["reason"].value_counts().to_string())

# Visual sanity check
fig, ax = plt.subplots(figsize=(8, 3.5))
im = ax.imshow(X_all[0], origin="lower", aspect="auto", cmap="magma")
ax.set_title(f"Log-Mel spectrogram — {meta.loc[0, 'file']} ({meta.loc[0, 'label']})")
ax.set_xlabel("Frame"); ax.set_ylabel("Mel band"); fig.colorbar(im, ax=ax, label="dB")
fig.tight_layout(); fig.savefig(FIG_DIR / "example_logmel.png", dpi=150); plt.show()
```

## 9. Patient-Independent Data Splitting

```python
# ============================================================
# 9. PATIENT-INDEPENDENT DATA SPLITTING (fixed for all runs)
# ============================================================
# Method: each patient has exactly one diagnosis, so the split is done at the PATIENT level and allocated
# class-by-class (i.e. stratified by patient diagnosis). For a class with n patients:
#   n >= 3 : n_val = n_test = max(1, round(0.15 n)), remainder -> train
#   n == 2 : 1 train, 0 val, 1 test      n == 1 : 1 train
# Percentages are therefore approximate at the recording level (patients differ greatly in recording count).
def patient_split(patients_df, seed, val_frac, test_frac):
    rng = np.random.RandomState(seed)
    assign = {}
    for label, grp in patients_df.groupby("label"):
        ids = np.array(sorted(grp["patient_id"]))
        rng.shuffle(ids)
        n = len(ids)
        if n >= 3:
            n_val = max(1, int(np.floor(val_frac * n + 0.5)))
            n_test = max(1, int(np.floor(test_frac * n + 0.5)))
        elif n == 2:
            n_val, n_test = 0, 1
        else:
            n_val, n_test = 0, 0
        for i, pid in enumerate(ids):
            assign[int(pid)] = "test" if i < n_test else ("val" if i < n_test + n_val else "train")
    return assign

patients = meta.drop_duplicates("patient_id")[["patient_id", "label"]]
assert patients["patient_id"].is_unique
assign = patient_split(patients, CFG["split_seed"], CFG["val_frac"], CFG["test_frac"])
meta["split"] = meta["patient_id"].map(assign)
assert meta["split"].notna().all()

# ---- Leakage check ----
ids = {s: set(meta.loc[meta["split"] == s, "patient_id"]) for s in ["train", "val", "test"]}
assert not (ids["train"] & ids["val"]) and not (ids["train"] & ids["test"]) and not (ids["val"] & ids["test"]), "PATIENT LEAKAGE!"
print("Leakage check passed: no patient appears in more than one split.\n")

for s in ["train", "val", "test"]:
    n_rec = int((meta["split"] == s).sum())
    print(f"{s.upper():5s}: {len(ids[s]):3d} patients ({100*len(ids[s])/len(patients):.1f}% of patients) | "
          f"{n_rec:4d} recordings ({100*n_rec/len(meta):.1f}% of recordings)")
    print("       patient IDs:", sorted(ids[s]))

pt_dist = pd.crosstab(meta.drop_duplicates("patient_id")["label"], meta.drop_duplicates("patient_id")["split"]).reindex(index=CLASS_NAMES, columns=["train", "val", "test"]).fillna(0).astype(int)
rec_dist = pd.crosstab(meta["label"], meta["split"]).reindex(index=CLASS_NAMES, columns=["train", "val", "test"]).fillna(0).astype(int)
print("\nPatients per class per split"); display(pt_dist)
print("Recordings per class per split"); display(rec_dist)

absent = {s: [c for c in CLASS_NAMES if rec_dist.loc[c, s] == 0] for s in ["train", "val", "test"]}
for s, a in absent.items():
    print(f"Classes absent from {s}: {a if a else 'none'}")
print("\nMetrics for classes absent from the validation / test set are reported as undefined; macro averages use only classes with support.")

meta[["file", "patient_id", "label", "split"]].to_csv(SPLIT_DIR / "split_assignment.csv", index=False)
pt_dist.to_csv(SPLIT_DIR / "patients_per_class_per_split.csv"); rec_dist.to_csv(SPLIT_DIR / "recordings_per_class_per_split.csv")
```

## 10. PyTorch Dataset and DataLoader

```python
# ============================================================
# 10. PYTORCH DATASET AND DATALOADER
# ============================================================
# Features live in CPU RAM (~150 MB); only mini-batches are moved to the GPU.
train_mask = (meta["split"] == "train").values
# Normalisation statistics come from the TRAINING split only (no val/test leakage)
FEAT_MEAN, FEAT_STD = float(X_all[train_mask].mean()), float(X_all[train_mask].std() + 1e-8)
print(f"Train-set feature mean={FEAT_MEAN:.3f}, std={FEAT_STD:.3f}")

X_t = torch.from_numpy((X_all - FEAT_MEAN) / FEAT_STD).unsqueeze(1)          # (N, 1, 128, 313)
y_t = torch.from_numpy(meta["label_idx"].values).long()
SPLIT_IDX = {s: torch.from_numpy(np.where(meta["split"].values == s)[0]) for s in ["train", "val", "test"]}
SPLIT_DS = {s: TensorDataset(X_t[idx], y_t[idx]) for s, idx in SPLIT_IDX.items()}
INPUT_SHAPE = tuple(X_t.shape[1:])
print("Input shape per sample:", INPUT_SHAPE, "| samples:", {s: len(d) for s, d in SPLIT_DS.items()})

def make_loader(split: str, seed: int, shuffle: bool = False) -> DataLoader:
    """Deterministic loader: shuffling order depends only on the run seed."""
    g = torch.Generator(); g.manual_seed(seed)
    return DataLoader(SPLIT_DS[split], batch_size=CFG["batch_size"], shuffle=shuffle, generator=g,
                      num_workers=0, pin_memory=(DEVICE.type == "cuda"), drop_last=False)
```

## 11. MobileNetV3 Architecture (scratch and pretrained variants)

```python
# ============================================================
# 11. MOBILENETV3 (1-channel spectrogram input) — "scratch" and "pretrained" variants
# ============================================================
from torchvision.models import (mobilenet_v3_large, MobileNet_V3_Large_Weights,
                                mobilenet_v3_small, MobileNet_V3_Small_Weights)

PRETRAINED_SOURCE = {"value": None}
_MB = {"large": (mobilenet_v3_large, MobileNet_V3_Large_Weights.IMAGENET1K_V1, "mobilenet_v3_large"),
       "small": (mobilenet_v3_small, MobileNet_V3_Small_Weights.IMAGENET1K_V1, "mobilenet_v3_small")}
MB_FN, MB_WEIGHTS, MB_NAME = _MB[CFG["mobilenet_size"]]

def load_imagenet_mobilenet():
    """ImageNet-pretrained torchvision MobileNetV3. Tries (1) torchvision download, (2) a local mobilenet_v3_*.pth found in /kaggle/input.
    It NEVER falls back silently to random weights: if neither works it raises with instructions."""
    try:
        m = MB_FN(weights=MB_WEIGHTS)
        PRETRAINED_SOURCE["value"] = f"torchvision download ({MB_NAME}, IMAGENET1K_V1)"
        return m
    except Exception as e:
        err = repr(e)
    key = MB_NAME.replace("_", "")
    local = sorted(p for p in INPUT_ROOT.rglob("*.pth") if key in p.name.lower().replace("_", "").replace("-", ""))
    if local:
        obj = torch.load(local[0], map_location="cpu")
        sd = obj.state_dict() if isinstance(obj, nn.Module) else (obj.get("state_dict", obj) if isinstance(obj, dict) else obj)
        m = MB_FN(weights=None); m.load_state_dict(sd)
        PRETRAINED_SOURCE["value"] = f"local file: {local[0]}"
        return m
    raise RuntimeError(f"Could not obtain ImageNet {MB_NAME} weights.\n"
                       f"  torchvision download failed with: {err}\n"
                       f"  Fix: (a) enable 'Internet' in the Kaggle notebook settings, or (b) add a Kaggle dataset that contains a "
                       f"{MB_NAME} *.pth ImageNet state_dict (file name containing '{MB_NAME}') and re-run.")

def build_model(variant: str, n_classes: int = 8, dropout: float = 0.5) -> nn.Module:
    """MobileNetV3 adapted to 1-channel input. All layers are trained (full fine-tuning) with the same optimizer/lr.
    variant='scratch'    : random initialisation.
    variant='pretrained' : ImageNet weights; the 3-channel stem conv is converted to 1 channel by summing its RGB filters."""
    if variant == "pretrained":
        m = load_imagenet_mobilenet()
    elif variant == "scratch":
        m = MB_FN(weights=None)
    else:
        raise ValueError(variant)
    old = m.features[0][0]                                       # stem Conv2d(3, C, 3x3, stride 2, bias=False)
    new = nn.Conv2d(1, old.out_channels, kernel_size=old.kernel_size, stride=old.stride, padding=old.padding, bias=False)
    if variant == "pretrained":
        with torch.no_grad():
            new.weight.copy_(old.weight.sum(dim=1, keepdim=True))
    else:
        nn.init.kaiming_normal_(new.weight, mode="fan_out", nonlinearity="relu")
    m.features[0][0] = new
    # classifier = [Linear(960|576 -> 1280|1024), Hardswish, Dropout(0.2), Linear(-> 1000)]  ->  keep hidden layer, Dropout(0.5), new 8-class head
    hidden = m.classifier[0].out_features
    m.classifier = nn.Sequential(m.classifier[0], m.classifier[1], nn.Dropout(dropout), nn.Linear(hidden, n_classes))
    return m

def count_params(m): return sum(p.numel() for p in m.parameters() if p.requires_grad)

set_seed(0)
_m = build_model("scratch", CFG["num_classes"])
print(_m)
print(f"\nTrainable parameters (both variants have the same architecture): {count_params(_m):,}")
with torch.no_grad():
    print("Output shape for a dummy batch:", tuple(_m.eval()(torch.zeros(2, *INPUT_SHAPE)).shape))
del _m
if "pretrained" in CFG["variants"]:
    _p = build_model("pretrained", CFG["num_classes"])
    print("Pretrained weights source:", PRETRAINED_SOURCE["value"])
    del _p
```

## 12. Training and Validation Functions

```python
# ============================================================
# 12. TRAINING AND VALIDATION FUNCTIONS
# ============================================================
def make_optimizer(name: str, params, lr: float):
    if name == "Adam":    return torch.optim.Adam(params, lr=lr)
    if name == "AdamW":   return torch.optim.AdamW(params, lr=lr)
    if name == "SGD":     return torch.optim.SGD(params, lr=lr, momentum=0.9)
    if name == "RMSprop": return torch.optim.RMSprop(params, lr=lr)
    if name == "NAdam":   return torch.optim.NAdam(params, lr=lr)
    if name == "RAdam":   return torch.optim.RAdam(params, lr=lr)
    if name == "Adamax":  return torch.optim.Adamax(params, lr=lr)
    if name == "Adagrad": return torch.optim.Adagrad(params, lr=lr)
    raise ValueError(name)

def train_one_epoch(model, loader, optimizer, criterion):
    model.train()
    total_loss, correct, n = 0.0, 0, 0
    for xb, yb in loader:
        xb, yb = xb.to(DEVICE, non_blocking=True), yb.to(DEVICE, non_blocking=True)
        optimizer.zero_grad(set_to_none=True)
        out = model(xb)
        loss = criterion(out, yb)
        loss.backward()
        optimizer.step()
        total_loss += loss.item() * xb.size(0)
        correct += (out.argmax(1) == yb).sum().item()
        n += xb.size(0)
    return total_loss / n, correct / n

@torch.no_grad()
def evaluate(model, loader, criterion):
    model.eval()
    total_loss, n = 0.0, 0
    ys, preds, probs = [], [], []
    for xb, yb in loader:
        xb, yb = xb.to(DEVICE, non_blocking=True), yb.to(DEVICE, non_blocking=True)
        out = model(xb)
        total_loss += criterion(out, yb).item() * xb.size(0)
        n += xb.size(0)
        ys.append(yb.cpu()); preds.append(out.argmax(1).cpu()); probs.append(torch.softmax(out, 1).cpu())
    y = torch.cat(ys).numpy(); p = torch.cat(preds).numpy(); pr = torch.cat(probs).numpy()
    return total_loss / n, float((y == p).mean()), y, p, pr

def to_cpu(obj):
    """Detached CPU copy of (nested) tensors, used to snapshot the best checkpoint."""
    if torch.is_tensor(obj): return obj.detach().cpu().clone()
    if isinstance(obj, dict): return {k: to_cpu(v) for k, v in obj.items()}
    if isinstance(obj, (list, tuple)): return type(obj)(to_cpu(v) for v in obj)
    return copy.deepcopy(obj)
```

## 13. Early Stopping Implementation

```python
# ============================================================
# 13. EARLY STOPPING (monitors validation Macro-F1, mode = max)
# ============================================================
class EarlyStopping:
    """Stops when the monitored score has not strictly improved for `patience` consecutive epochs.
    Keeps an in-memory snapshot (model + optimizer state) of the best epoch."""
    def __init__(self, patience: int):
        self.patience, self.best_score, self.best_epoch = patience, -np.inf, 0
        self.counter, self.should_stop = 0, False
        self.best_model_state, self.best_opt_state = None, None
    def step(self, score, epoch, model, optimizer) -> bool:
        if score > self.best_score:
            self.best_score, self.best_epoch, self.counter = score, epoch, 0
            self.best_model_state = to_cpu(model.state_dict())   # model weights only (no checkpoint is written to disk)
            return True
        self.counter += 1
        self.should_stop = self.counter >= self.patience
        return False
```

## 14. Evaluation Metrics (and the single-experiment runner)

```python
# ============================================================
# 14. EVALUATION METRICS + EXPERIMENT RUNNER
# ============================================================
def compute_metrics(y_true, y_pred, y_prob=None, n_classes=8):
    """All metrics derived from the 8x8 confusion matrix.
    - Class-wise precision is undefined (NaN) if the class was never predicted; recall is undefined if the class has no true samples.
    - Macro averages use ONLY classes with support>0 in y_true (precision of a supported-but-never-predicted class counts as 0).
    - Macro AUC (one-vs-rest) averages the per-class AUCs that are mathematically defined (class present AND at least one other class present)."""
    cm = confusion_matrix(y_true, y_pred, labels=list(range(n_classes)))
    tp = np.diag(cm).astype(float); sup = cm.sum(1).astype(float); pc = cm.sum(0).astype(float)
    fp, fn = pc - tp, sup - tp
    with np.errstate(divide="ignore", invalid="ignore"):
        prec = np.where(pc > 0, tp / pc, np.nan)
        rec = np.where(sup > 0, tp / sup, np.nan)
        den = 2 * tp + fp + fn
        f1 = np.where(den > 0, 2 * tp / den, np.nan)
    present = sup > 0
    out = dict(accuracy=float(tp.sum() / cm.sum()),
               macro_precision=float(np.mean(np.nan_to_num(prec[present], nan=0.0))),
               macro_recall=float(np.mean(rec[present])),
               macro_f1=float(np.mean(f1[present])),
               n_classes_evaluated=int(present.sum()), cm=cm, prec=prec, rec=rec, f1=f1, support=sup)
    aucs = np.full(n_classes, np.nan)
    if y_prob is not None:
        for c in range(n_classes):
            pos = (np.asarray(y_true) == c)
            if 0 < pos.sum() < len(pos):
                aucs[c] = roc_auc_score(pos, y_prob[:, c])
    out["auc_per_class"] = aucs
    out["macro_auc"] = float(np.nanmean(aucs)) if np.isfinite(aucs).any() else float("nan")
    out["n_auc_classes"] = int(np.isfinite(aucs).sum())
    return out

def run_experiment(opt_name: str, variant: str, lr: float, seed: int) -> dict:
    run_id = f"{opt_name}_{variant}_lr{lr:g}_seed{seed}"
    rdir = RUNS_DIR / run_id
    if (rdir / "result.json").exists():                          # resume support: skip finished runs
        return json.load(open(rdir / "result.json"))
    rdir.mkdir(parents=True, exist_ok=True)

    set_seed(seed)
    model = build_model(variant, CFG["num_classes"]).to(DEVICE)  # fresh model for every run
    optimizer = make_optimizer(opt_name, model.parameters(), lr)
    criterion = nn.CrossEntropyLoss()                            # no class weights (nothing unreported)
    tr_loader, va_loader, te_loader = make_loader("train", seed, True), make_loader("val", seed), make_loader("test", seed)
    stopper, history = EarlyStopping(CFG["patience"]), []

    t0 = time.time()
    for epoch in range(1, CFG["max_epochs"] + 1):
        tr_loss, tr_acc = train_one_epoch(model, tr_loader, optimizer, criterion)
        va_loss, va_acc, yv, pv, _ = evaluate(model, va_loader, criterion)
        vm = compute_metrics(yv, pv, None, CFG["num_classes"])
        history.append(dict(optimizer=opt_name, model_variant=variant, learning_rate=lr, seed=seed, epoch=epoch,
                            train_loss=tr_loss, train_acc=tr_acc, val_loss=va_loss, val_acc=va_acc, val_macro_f1=vm["macro_f1"]))
        stopper.step(vm["macro_f1"], epoch, model, optimizer)
        if stopper.should_stop: break
    train_time = time.time() - t0

    # ---- restore best checkpoint (selected on validation Macro-F1) and evaluate ONCE on the held-out test set ----
    model.load_state_dict(stopper.best_model_state)
    te_loss, _, yt, pt, prt = evaluate(model, te_loader, criterion)
    tm = compute_metrics(yt, pt, prt, CFG["num_classes"])
    best_val_row = history[stopper.best_epoch - 1]

    # (no checkpoint is saved: keeps the output small)
    pd.DataFrame(history).to_csv(rdir / "history.csv", index=False)
    pd.DataFrame(dict(optimizer=opt_name, model_variant=variant, learning_rate=lr, seed=seed, **{"class": CLASS_NAMES},
                      support=tm["support"].astype(int), precision=tm["prec"], recall=tm["rec"], f1=tm["f1"],
                      auc=tm["auc_per_class"])).to_csv(rdir / "classwise.csv", index=False)
    np.save(rdir / "confusion_matrix.npy", tm["cm"])

    result = dict(run_id=run_id, optimizer=opt_name, model="mobilenetv3", model_variant=variant, learning_rate=lr, seed=seed,
                  best_epoch=stopper.best_epoch, epochs_run=len(history), stopped_early=bool(stopper.should_stop),
                  best_val_macro_f1=float(stopper.best_score), val_accuracy_at_best=float(best_val_row["val_acc"]),
                  test_accuracy=tm["accuracy"], test_macro_precision=tm["macro_precision"], test_macro_recall=tm["macro_recall"],
                  test_macro_f1=tm["macro_f1"], test_macro_auc=tm["macro_auc"], test_loss=float(te_loss),
                  n_test_classes_evaluated=tm["n_classes_evaluated"], n_auc_classes=tm["n_auc_classes"],
                  training_time_s=float(train_time), n_params=count_params(model))
    json.dump(result, open(rdir / "result.json", "w"), indent=1)      # written last = marks the run as complete

    del model, optimizer, tr_loader, va_loader, te_loader, stopper
    gc.collect()
    if DEVICE.type == "cuda": torch.cuda.empty_cache()
    return result

def run_optimizer_block(opt_name: str):
    """Runs (or skips if finished) every model-variant x learning-rate x seed combination for one optimizer."""
    jobs = [(v, lr, s) for v in CFG["variants"] for lr in CFG["learning_rates"] for s in CFG["seeds"]]
    results = []
    for v, lr, s in tqdm(jobs, desc=f"{opt_name} runs"):
        r = run_experiment(opt_name, v, lr, s)
        results.append(r)
        tqdm.write(f"{r['run_id']:38s} best_ep={r['best_epoch']:3d}/{r['epochs_run']:3d} val_F1={r['best_val_macro_f1']:.4f} "
                   f"| test acc={r['test_accuracy']:.4f} F1={r['test_macro_f1']:.4f} AUC={r['test_macro_auc']:.4f} | {r['training_time_s']:.0f}s")
    return pd.DataFrame(results)
```

## 15. Run All 8 Optimizers

```python
# ============================================================
# 15. RUN ALL 8 OPTIMIZERS  (8 x 2 variants x 2 learning rates x 5 seeds = 160 runs; finished runs are skipped)
#     After each optimizer finishes, its own CSV is written to RES_DIR.
# ============================================================
MAIN_COLS = {"optimizer": "Optimizer", "model_variant": "Model Variant", "learning_rate": "Learning Rate", "seed": "Seed", "best_epoch": "Best Epoch",
             "best_val_macro_f1": "Best Val Macro-F1", "test_accuracy": "Accuracy", "test_macro_precision": "Macro Precision",
             "test_macro_recall": "Macro Recall", "test_macro_f1": "Macro-F1", "test_macro_auc": "Macro AUC",
             "training_time_s": "Training Time (s)", "n_params": "Number of Parameters",
             "epochs_run": "Epochs Run", "n_test_classes_evaluated": "Test Classes With Support", "n_auc_classes": "Classes In Macro AUC"}

def load_all_results() -> pd.DataFrame:
    rows = [json.load(open(p)) for p in sorted(RUNS_DIR.glob("*/result.json"))]
    return pd.DataFrame(rows)

def export_optimizer_csv(opt: str) -> pd.DataFrame:
    """One CSV per optimizer: one row per finished run (variant x lr x seed)."""
    df = load_all_results()
    df = df[df["optimizer"] == opt].sort_values(["model_variant", "learning_rate", "seed"]).reset_index(drop=True)
    out = df[list(MAIN_COLS)].rename(columns=MAIN_COLS)
    out.to_csv(RES_DIR / f"mobilenetv3_{opt.lower()}_results.csv", index=False)
    return out

# ---- PRE-FLIGHT CHECK: one batch, both variants, Adam. Shows clearly what fails BEFORE the long run starts ----
import traceback
def preflight():
    xb, yb = next(iter(make_loader("train", 0, True)))
    for v in CFG["variants"]:
        try:
            set_seed(0)
            m = build_model(v, CFG["num_classes"]).to(DEVICE)
            o = make_optimizer("Adam", m.parameters(), 1e-3)
            m.train()
            loss = nn.CrossEntropyLoss()(m(xb.to(DEVICE)), yb.to(DEVICE))
            o.zero_grad(); loss.backward(); o.step()
            print(f"[preflight] {v:10s} OK  (loss={loss.item():.3f}, batch={tuple(xb.shape)}, device={DEVICE})")
            del m, o
        except Exception:
            print(f"[preflight] {v} FAILED:"); traceback.print_exc()
            if v == "pretrained":
                print("\n>>> Pretrained MobileNetV3 weights could not be loaded. Fix: Kaggle Settings -> Internet = On, "
                      "or add a dataset that contains a mobilenet_v3_large*.pth ImageNet file.")
            return False
    return True

if not preflight():
    raise RuntimeError("Pre-flight check failed - read the message above (nothing was trained).")

failed = []
for opt in CFG["optimizers"]:
    print(f"\n{'=' * 18} {opt} {'=' * 18}")
    try:
        res = run_optimizer_block(opt)
        export_optimizer_csv(opt)
    except Exception:
        print(f"[{opt}] FAILED - full error below; continuing with the next optimizer.")
        traceback.print_exc(); failed.append(opt)
        continue
    print(f"[{opt}] CSV saved -> {RES_DIR / f'mobilenetv3_{opt.lower()}_results.csv'}")
    print(f"[{opt}] mean test Macro-F1 per variant/LR:")
    display(res.groupby(["model_variant", "learning_rate"])[["best_val_macro_f1", "test_accuracy", "test_macro_f1", "test_macro_auc"]].mean().round(4))

print("\nFailed optimizers:", failed if failed else "none")
```

## 16. Confusion Matrices and Training Curves

```python
# ============================================================
# 16. CONFUSION MATRICES AND TRAINING CURVES  (main ones only)
#     For every optimizer x variant, only the BEST run (highest VALIDATION Macro-F1) gets a confusion matrix + training curve.
# ============================================================
results_df = load_all_results()
expected = len(CFG["optimizers"]) * len(CFG["variants"]) * len(CFG["learning_rates"]) * len(CFG["seeds"])
print(f"Completed runs found: {len(results_df)} / {expected}")
if len(results_df) < expected:
    print("WARNING: some runs are missing - re-run section 15 to finish them.")

def plot_confusion(cm, title, path, show=False):
    support = cm.sum(1)
    norm = np.divide(cm, support[:, None], out=np.zeros(cm.shape, dtype=float), where=support[:, None] > 0)
    fig, ax = plt.subplots(figsize=(8.5, 7))
    im = ax.imshow(norm, cmap="Blues", vmin=0, vmax=1)
    ax.set_xticks(range(len(CLASS_NAMES))); ax.set_xticklabels(CLASS_NAMES, rotation=45, ha="right")
    ax.set_yticks(range(len(CLASS_NAMES))); ax.set_yticklabels([f"{c} (n={int(s)})" for c, s in zip(CLASS_NAMES, support)])
    for i in range(cm.shape[0]):
        for j in range(cm.shape[1]):
            ax.text(j, i, int(cm[i, j]), ha="center", va="center", color="white" if norm[i, j] > 0.5 else "black")
    ax.set_xlabel("Predicted label"); ax.set_ylabel("True label (n = test recordings)")
    ax.set_title(title); fig.colorbar(im, ax=ax, label="Row-normalised fraction")
    fig.tight_layout(); fig.savefig(path, dpi=150)
    if show: plt.show()
    plt.close(fig)

def plot_curves(hist: pd.DataFrame, title, path, show=False):
    fig, axes = plt.subplots(2, 2, figsize=(11, 7))
    spec = [("train_loss", "Training Loss vs Epoch", "Loss"), ("val_loss", "Validation Loss vs Epoch", "Loss"),
            ("train_acc", "Training Accuracy vs Epoch", "Accuracy"), ("val_acc", "Validation Accuracy vs Epoch", "Accuracy")]
    best_ep = hist.loc[hist["val_macro_f1"].idxmax(), "epoch"]
    for ax, (col, ttl, yl) in zip(axes.ravel(), spec):
        ax.plot(hist["epoch"], hist[col], label=col.replace("_", " "))
        ax.axvline(best_ep, color="red", ls="--", alpha=0.6, label=f"best epoch ({int(best_ep)})")
        ax.set_title(ttl); ax.set_xlabel("Epoch"); ax.set_ylabel(yl); ax.legend(); ax.grid(alpha=0.3)
    fig.suptitle(title); fig.tight_layout(); fig.savefig(path, dpi=150)
    if show: plt.show()
    plt.close(fig)

best_runs = results_df.loc[results_df.groupby(["optimizer", "model_variant"])["best_val_macro_f1"].idxmax()]
overall_best_ids = results_df.loc[results_df.groupby("model_variant")["best_val_macro_f1"].idxmax(), "run_id"].tolist()   # only these are displayed inline
for r in tqdm(best_runs.to_dict("records"), desc="Saving main figures"):
    rdir = RUNS_DIR / r["run_id"]
    d = FIG_DIR / "per_optimizer" / r["optimizer"]; d.mkdir(parents=True, exist_ok=True)
    show = r["run_id"] in overall_best_ids
    ttl = f"MobileNetV3 ({r['model_variant']}) + {r['optimizer']} lr={r['learning_rate']:g} seed={r['seed']}"
    plot_confusion(np.load(rdir / "confusion_matrix.npy"),
                   f"Confusion matrix (test) — {ttl}\nacc={r['test_accuracy']:.3f}, macro-F1={r['test_macro_f1']:.3f}",
                   d / f"cm_{r['model_variant']}.png", show=show)
    plot_curves(pd.read_csv(rdir / "history.csv"), f"Training curves — {ttl}", d / f"curves_{r['model_variant']}.png", show=show)
print("Saved figures for", len(best_runs), "best runs (one per optimizer x variant).")
print("Displayed inline: best run per variant overall ->", overall_best_ids)
```

## 18. Final Results: Optimizer Comparison

```python
# ============================================================
# 18. FINAL RESULTS: optimizer comparison (selection on validation Macro-F1 only)
# ============================================================
opts, variants = CFG["optimizers"], CFG["variants"]

# For every optimizer x variant pick the learning rate with the higher mean VALIDATION Macro-F1
best_tbl = summary.loc[summary.groupby(["optimizer", "model_variant"])["best_val_macro_f1_mean"].idxmax()].reset_index(drop=True)
comparison = best_tbl[["optimizer", "model_variant", "learning_rate", "n_seeds",
                       "best_val_macro_f1_mean", "best_val_macro_f1_std", "test_accuracy_mean", "test_accuracy_std",
                       "test_macro_precision_mean", "test_macro_recall_mean", "test_macro_f1_mean", "test_macro_f1_std",
                       "test_macro_auc_mean", "test_macro_auc_std", "training_time_s_mean"]]
comparison = comparison.sort_values("best_val_macro_f1_mean", ascending=False).reset_index(drop=True)
comparison.to_csv(WORK_RES / "final_optimizer_comparison.csv", index=False)
print("Final comparison (each optimizer x variant at its best LR by validation Macro-F1; test metrics = mean ± SD over seeds)")
display(comparison.round(4))

top = comparison.iloc[0]
print("\n=== Best configuration (selected by mean validation Macro-F1) ===")
print(f"MobileNetV3 ({top['model_variant']}) + {top['optimizer']}, LR={top['learning_rate']:g}")
print(f"  Validation Macro-F1 : {top['best_val_macro_f1_mean']:.4f} ± {top['best_val_macro_f1_std']:.4f}")
for lab, k in [("Accuracy", "test_accuracy"), ("Macro-F1", "test_macro_f1"), ("Macro AUC", "test_macro_auc")]:
    print(f"  Held-out test {lab:10s}: {top[k + '_mean']:.4f} ± {top[k + '_std']:.4f}")

# ---- Plot 1: optimizer comparison (test Macro-F1, test accuracy, validation vs test) ----
fig, axes = plt.subplots(1, 3, figsize=(20, 5))
w = 0.8 / len(variants); x = np.arange(len(opts))
for k, v in enumerate(variants):
    sub = best_tbl[best_tbl["model_variant"] == v].set_index("optimizer").reindex(opts)
    axes[0].bar(x + k * w, sub["test_macro_f1_mean"], w, yerr=sub["test_macro_f1_std"], capsize=3, label=v)
    axes[1].bar(x + k * w, sub["test_accuracy_mean"], w, yerr=sub["test_accuracy_std"], capsize=3, label=v)
    axes[2].scatter(sub["best_val_macro_f1_mean"], sub["test_macro_f1_mean"], s=70, label=v)
    for o in opts:
        if np.isfinite(sub.loc[o, "test_macro_f1_mean"]):
            axes[2].annotate(o, (sub.loc[o, "best_val_macro_f1_mean"], sub.loc[o, "test_macro_f1_mean"]), fontsize=8, xytext=(3, 3), textcoords="offset points")
for a, ttl, yl in [(axes[0], "Test Macro-F1 (mean ± SD over seeds, best LR)", "Macro-F1"), (axes[1], "Test Accuracy (mean ± SD over seeds, best LR)", "Accuracy")]:
    a.set_xticks(x + w * (len(variants) - 1) / 2); a.set_xticklabels(opts, rotation=30, ha="right")
    a.set_title(ttl); a.set_ylabel(yl); a.set_xlabel("Optimizer"); a.legend(); a.grid(axis="y", alpha=0.3)
axes[2].set_title("Validation vs test Macro-F1 (mean)"); axes[2].set_xlabel("Validation Macro-F1"); axes[2].set_ylabel("Test Macro-F1"); axes[2].legend(); axes[2].grid(alpha=0.3)
fig.tight_layout(); fig.savefig(FIG_DIR / "optimizer_comparison.png", dpi=150); plt.show()

# ---- Plot 2: heat-map of mean test Macro-F1 (optimizer x variant/learning rate) ----
pv = summary.pivot_table(index="optimizer", columns=["model_variant", "learning_rate"], values="test_macro_f1_mean").reindex(opts)
fig, ax = plt.subplots(figsize=(9, 5.5))
im = ax.imshow(pv.values, cmap="viridis", aspect="auto")
ax.set_xticks(range(pv.shape[1])); ax.set_xticklabels([f"{v}\nlr={lr:g}" for v, lr in pv.columns])
ax.set_yticks(range(pv.shape[0])); ax.set_yticklabels(pv.index)
vmid = np.nanmean(pv.values)
for i in range(pv.shape[0]):
    for j in range(pv.shape[1]):
        if np.isfinite(pv.values[i, j]):
            ax.text(j, i, f"{pv.values[i, j]:.3f}", ha="center", va="center", color="white" if pv.values[i, j] < vmid else "black")
ax.set_title("Mean test Macro-F1 — optimizer × variant / learning rate"); fig.colorbar(im, ax=ax, label="Macro-F1")
fig.tight_layout(); fig.savefig(FIG_DIR / "optimizer_lr_heatmap.png", dpi=150); plt.show()

# ---- Objective, data-driven reading of the results (no claims of significance) ----
print("\n=== Objective summary (generated from the numbers above) ===")
for i, r in comparison.head(5).iterrows():
    print(f"{i+1}. {r['optimizer']:8s} {r['model_variant']:10s} lr={r['learning_rate']:<7g} val F1={r['best_val_macro_f1_mean']:.3f} | test F1={r['test_macro_f1_mean']:.3f} ± {r['test_macro_f1_std']:.3f}")
print(f"\n- Spread of mean test Macro-F1 across optimizer/variant configurations: {comparison['test_macro_f1_mean'].min():.3f} – {comparison['test_macro_f1_mean'].max():.3f}.")
if len(comparison) > 1 and comparison.iloc[1]["test_macro_f1_mean"] + comparison.iloc[1]["test_macro_f1_std"] >= top["test_macro_f1_mean"] - top["test_macro_f1_std"]:
    print("- The ±1 SD ranges of the top two configurations overlap, so their order should be treated as unresolved.")
for v in variants:
    print(f"- Mean test Macro-F1 over all optimizers/LRs for '{v}': {summary[summary['model_variant'] == v]['test_macro_f1_mean'].mean():.3f}")
print("- Caveats: (i) the split is fixed, so seed variance covers initialisation/shuffling only, not patient sampling; "
      "(ii) rare classes have very few patients, so validation/test metrics rest on a handful of individuals and classes absent "
      "from a split are excluded from macro averages; (iii) no statistical significance test was run, so differences are descriptive only.")
```

## 19. Zip the Outputs and Split into 5 Parts for Download

```python
# ============================================================
# 19. ZIP THE OUTPUTS AND SPLIT THE ZIP INTO 5 PARTS FOR DOWNLOAD
#     Zip = the 8 CSV files + all main figures. Only the 5 parts (and the 8 CSVs) stay in /kaggle/working.
# ============================================================
import zipfile, math

zip_path = WORK / "mobilenetv3_all_results.zip"              # temp location (the full zip is not part of the output)
with zipfile.ZipFile(zip_path, "w", zipfile.ZIP_DEFLATED) as z:
    for p in sorted(RES_DIR.glob("*.csv")):
        z.write(p, Path("results") / p.name)
    for p in sorted(FIG_DIR.rglob("*.png")):
        z.write(p, Path("figures") / p.relative_to(FIG_DIR))
total_size = zip_path.stat().st_size
n_files = len(zipfile.ZipFile(zip_path).namelist())

parts = 5
for old in DL_DIR.glob("*"):                              # clean a previous run
    old.unlink()
chunk_size = max(1, math.ceil(total_size / parts))
print(f"Zip: {n_files} files | {total_size / 1024**2:.2f} MB | {parts} parts of about {chunk_size / 1024**2:.2f} MB")

part_names = []
with open(zip_path, "rb") as source:
    for i in range(1, parts + 1):
        data = source.read(chunk_size)
        name = f"mobilenetv3_all_part{i}.bin"
        with open(DL_DIR / name, "wb") as dst:
            dst.write(data)
        part_names.append(name)
        print(f"Part {i}/{parts} completed: {len(data) / 1024**2:.2f} MB")
assert sum((DL_DIR / n).stat().st_size for n in part_names) == total_size

howto = (
    "HOW TO REJOIN THE 5 PARTS INTO THE ZIP\n"
    "--------------------------------------\n"
    "1) Download all 5 files (mobilenetv3_all_part1.bin ... mobilenetv3_all_part5.bin) into the SAME folder.\n"
    "2) Windows (Command Prompt, inside that folder):\n"
    "   copy /b mobilenetv3_all_part1.bin+mobilenetv3_all_part2.bin+mobilenetv3_all_part3.bin+mobilenetv3_all_part4.bin+mobilenetv3_all_part5.bin mobilenetv3_all_results.zip\n"
    "3) Linux / macOS:\n"
    "   cat mobilenetv3_all_part1.bin mobilenetv3_all_part2.bin mobilenetv3_all_part3.bin mobilenetv3_all_part4.bin mobilenetv3_all_part5.bin > mobilenetv3_all_results.zip\n"
    "4) Unzip mobilenetv3_all_results.zip -> results/ (8 CSV files) and figures/ (PNG graphs).\n"
)
(DL_DIR / "HOW_TO_REJOIN.txt").write_text(howto)

work_files = [p for p in OUT_ROOT.rglob("*") if p.is_file()]
print(f"\nAll parts created successfully! Folder: {DL_DIR}")
print(f"Files now in {OUT_ROOT}: {len(work_files)} | total size: {sum(p.stat().st_size for p in work_files) / 1024**2:.1f} MB")
print("\n" + howto)
```
